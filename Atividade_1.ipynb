```python
def realizar_saque(saldo_atual, valor_saque):
    # Verifica se o cliente tem saldo
    if valor_saque > saldo_atual:
        return "Erro: Saldo insuficiente"
    else:
        novo_saldo = saldo_atual - valor_saque
        return novo_saldo
```

```python
import unittest

class TestCaixaEletronico(unittest.TestCase):
    def test_saque_caminho_feliz(self):
        resultado = realizar_saque(100, 40)
        self.assertEqual(resultado, 60)

    def test_saque_saldo_insuficiente(self):
        resultado = realizar_saque(100, 150)
        self.assertEqual(resultado, "Erro: Saldo insuficiente")

    def test_saque_valor_negativo(self):
        resultado = realizar_saque(100, -20)
        self.assertEqual(resultado, "Erro: Valor inválido")

    def test_saque_valor_zerado(self):
        resultado = realizar_saque(100, 0)
        self.assertEqual(resultado, "Erro: Valor inválido")

# Comando adaptado para rodar o unittest dentro do Google Colab
if __name__ == '__main__':
    unittest.main(argv=['first-arg-is-ignored'], exit=False)
```

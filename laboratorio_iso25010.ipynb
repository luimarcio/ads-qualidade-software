### Cenário
Você foi contratado como Analista de Qualidade de Software para avaliar o backend de um sistema de processamento de dados financeiros. A função principal atual recebe uma lista de milhares de transações e remove as duplicadas.

No entanto, o sistema está apresentando lentidão extrema em produção, travando quando recebe dados inesperados e os novos desenvolvedores reclamam que o código é difícil de entender.

**Seu objetivo:** Avaliar e refatorar o código abaixo utilizando três características do Modelo de Qualidade do Produto da **ISO/IEC 25010**:
1. **Eficiência de Desempenho** (Comportamento no tempo)
2. **Confiabilidade** (Tolerância a falhas)
3. **Manutenibilidade** (Analisabilidade)

```python
import time
import random

# Gerando um grande volume de dados simulados (15.000 transações com repetições)
dados_simulados = [random.randint(1, 5000) for _ in range(15000)]

# ==========================================
# CÓDIGO LEGADO - NÃO ALTERAR ESTA CÉLULA
# ==========================================
def proc(d):
    r = []
    for i in range(len(d)):
        dup = False
        for j in range(len(r)):
            if d[i] == r[j]:
                dup = True
        if not dup:
            r.append(d[i])
    return r

# Teste de Desempenho do Código Legado
inicio = time.time()
resultado_legado = proc(dados_simulados)
fim = time.time()

tempo_legado = fim - inicio
print(f"Tempo de execução (Legado): {tempo_legado:.4f} segundos")
print(f"Total de itens únicos: {len(resultado_legado)}")
```
